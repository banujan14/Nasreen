Dictionary

```python
patient_record = {
    'name': 'John Doe',
    'age': 30,
    'conditions': ['diabetes', 'hypertension'],
    'medications': {
        'diabetes': 'metformin',
        'hypertension': 'lisinopril'
    }
}
print(patient_record)

print(patient_record['medications']['diabetes'])
print(patient_record['conditions'][1])
```

STACK-LIFO

```python
browser_history = ['irctc.com', 'yahoomail.com', 'github.com']
print(browser_history)
browser_history.append('python.org')
print(browser_history)
last_page = browser_history.pop()
print(f"Last visited page: {last_page}")
print(browser_history)
```

QUEUE-FIFO

```python
from collections import deque
ticket_queue = deque()
ticket_queue.append('Ticket 1') ## Enqueue
ticket_queue.append('Ticket 2')
print(ticket_queue)
served = ticket_queue.popleft() ## Deque
print(f"Serving: {served}")
```

NAMED TUPLE

```python
from collections import namedtuple
transaction = namedtuple('Transaction', ['customerid', 'amount', 'txntype'])
t1 = transaction(customerid=1, amount=100.0, txntype='debit')
print(t1)
```
